```python
'''
Bài 1: Xử lý ngoại lệ FileNotFoundError:
Lỗi này xuất hiện khi mở 1 file mà không tồn tại. Không tồn tại có thể do file chưa được tạo hoặc đường dẫn đến file không đúng.
'''
filename = 'alice23.txt'
try:
    with open(filename) as f_obj:
         contents = f_obj.read()
except FileNotFoundError:
     msg = f"File {filename} không tồn tại."
     print(msg)
```

```python
'''
Bài 2: về xử lý ngoại lệ FileNotFoundError khi thực hiện chương trình đếm số từ trong file text
'''
fileName = 'alice.txt'
try:
    with open(fileName) as f:
        contents = f.read()
except FileNotFoundError:
    msg = 'File ' + fileName + ' khong ton tai.'
    print(msg)
else:
    words = contents.split()
    num_words = len(words)
    print(f"File {fileName} có {num_words} từ.")
```

```python
''' 
Bài 3: về thực hiện chương trình đếm số từ trong nhiều file text
'''
def count_words(fileName):
    try:
        with open(fileName, encoding='utf-8') as f:
            contents = f.read()
    except FileNotFoundError:
        msg = f'File {fileName} khong ton tai.'
        print(msg)
    else:
        words = contents.split()
        num_words = len(words)
        print(f"File {fileName} có {num_words} từ.")
        
filenames = ['alice.txt', 'tuyenngondoclap.txt', 'f3.txt']
for filename in filenames:
     count_words(filename)                   
```

```python
'''
Bài 4. Tệp nén là một loại lưu trữ chứa một hoặc nhiều tệp đã bị giảm kích thước. Nén tập tin trong các hệ điều hành hiện đại thường khá đơn giản. 
Viết chương trình nén và giải nén tệp bằng Python
'''
import os
import zipfile
from pathlib import Path

def compress_zip(input_paths, zip_path):
    """
    Nén danh sách file/thư mục vào một file .zip
    input_paths: list[str | Path]  (file hoặc thư mục)
    zip_path: str | Path           (đường dẫn .zip sẽ tạo)
    """
    zip_path = Path(zip_path)
    with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
        for p in map(Path, input_paths):
            if p.is_file():
                # arcname để bỏ phần đường dẫn gốc, chỉ lưu tên tương đối
                zf.write(p, arcname=p.name)
            elif p.is_dir():
                for root, _, files in os.walk(p):
                    root_path = Path(root)
                    for name in files:
                        full = root_path / name
                        # lưu đường dẫn tương đối so với thư mục gốc
                        arcname = full.relative_to(p.parent)
                        zf.write(full, arcname=arcname)
            else:
                print(f"[BỎ QUA] Không tìm thấy: {p}")

    print(f"Đã nén thành công → {zip_path.resolve()}")

def decompress_zip(zip_path, extract_dir):
    """
    Giải nén file .zip vào thư mục đích
    """
    zip_path = Path(zip_path)
    extract_dir = Path(extract_dir)
    extract_dir.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path, "r") as zf:
        zf.extractall(extract_dir)
    print(f"Đã giải nén → {extract_dir.resolve()}")

# --- Ví dụ dùng ---
if __name__ == "__main__":
    # Nén nhiều tệp/thư mục
    compress_zip(["nen_file_zip.txt", "docs"], "archive.zip")

    # Giải nén
    decompress_zip("archive.zip", "extracted_zip")

```
