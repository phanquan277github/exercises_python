```python
'''
Bài 6: Viết chương trình tính thể tích hình trụ V = πr2h 
với r là bán kính của mặt đáy, h là chiều cao của hình trụ, và π là hằng số pi. 
Với r và h nhập vào từ bàn phím.
'''

import math
r = float(input("Nhập bán kính r: "))
h = float(input("Nhập chiều cao h: "))

V = math.pi * (r ** 2) * h

print("Thể tích hình trụ là:", V)
```

```python
'''
Bài 7: Viết chương trình tính diện tích tam giác theo công thức. Yêu cầu nhập vào 3 cạnh và các góc A,B và C (radian)
S=(1/2)ab*sin(c)=(1/2)ac*sin(b)=(1/2)bc*sin(a)
'''

import math

a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))
c = float(input("Nhập cạnh c: "))

A = math.radians(float(input("Nhập góc A (độ): ")))
B = math.radians(float(input("Nhập góc B (độ): ")))
C = math.radians(float(input("Nhập góc C (độ): ")))

S1 = 0.5 * a * b * math.sin(C)
S2 = 0.5 * a * c * math.sin(B)
S3 = 0.5 * b * c * math.sin(A)

print("Diện tích tam giác theo công thức 1:", round(S1, 3))
print("Diện tích tam giác theo công thức 2:", round(S2, 3))
print("Diện tích tam giác theo công thức 3:", round(S3, 3))
```

```python
'''
Bài 12: Viết chương trình nhập chi phí của một bữa ăn tại nhà hàng từ người dùng.
Chương trình sẽ tính thuế và tiền boa cho bữa ăn. Trong đó, tiền boa là 18% số tiền bữa ăn (không có thuế); tiền thuế là 5% số tiền bữa ăn. 
Đầu ra của chương trình là tổng tiền, gồm: thuế, số tiền boa và tiền bữa ăn. 
Định dạng đầu ra sao cho tất cả các giá trị được hiển thị bằng hai con số thập phân
'''

meal_cost = float(input("Nhập chi phí bữa ăn: "))

tip = meal_cost * 0.18
tax = meal_cost * 0.05
total = meal_cost + tip + tax

print(f"Tiền bữa ăn: {meal_cost:.2f}")
print(f"Tiền boa (18%): {tip:.2f}")
print(f"Thuế (5%): {tax:.2f}")
print(f"Tổng cộng: {total:.2f}")
```

```python
'''
Bài 14: Viết một chương trình đọc hai số nguyên a và b từ người dùng. Chương trình sẽ tính toán và hiển thị:
Tổng của a và b
Hiệu của a và b
Tích của a và b
Thương của a chia cho b
Phần còn lại khi a được chia cho b
Kết quả của log_10a
Kết quả của ab
'''
import math

a = int(input("Nhập số nguyên a: "))
b = int(input("Nhập số nguyên b: "))

tong = a + b
hieu = a - b
tich = a * b
thuong = a / b if b != 0 else "Không chia được cho 0"
chia_du = a % b if b != 0 else "Không chia được cho 0"
log_a = math.log10(a) if a > 0 else "Không tính được log10(a) khi a ≤ 0"
mu = a ** b

print("Tổng của a và b:", tong)
print("Hiệu của a và b:", hieu)
print("Tích của a và b:", tich)
print("Thương của a / b:", thuong)
print("Phần còn lại khi a / b:", chia_du)
print("Kết quả log10(a):", log_a)
print("Kết quả a^b:", mu)
```

```python
'''
'\n Bài 15: Năng lượng Q cần để thay đổi ΔT nhiệt độ (xét trường hợp ΔT > 0) của M gram nước  được tính bằng: Q = M × C × ΔT, trong đó ΔT có đơn vị là độ (độ Celsius), M có đơn vị gram và Q có đơn vị là Joules.  Viết chương trình theo yêu cầu:
Đọc khối lượng của nước và sự thay đổi ΔT nhiệt độ từ người dùng. Chương trình sẽ hiển thị tổng năng lượng của nước khi thay đổi nhiệt độ mong muốn. 
Gợi ý: Nhiệt dung riêng của nước là 4.186 J/g◦C.
Tính toán chi phí khi thay đổi ΔT nhiệt độ của nước, biết rằng: chi phí là 8,9 cent mỗi kilowatt giờ, 1 Joules = 2.777e-7 kWh.  Hãy tính chi phí để nước đạt nhiệt độ như yêu cầu.
'''
C = 4.186          
cost_per_kWh = 8.9 
J_to_kWh = 2.777e-7


M = float(input("Nhập khối lượng nước (gram): "))
delta_T = float(input("Nhập sự thay đổi nhiệt độ ΔT (°C): "))

Q = M * C * delta_T

Q_kWh = Q * J_to_kWh

cost = Q_kWh * cost_per_kWh

print(f"Năng lượng cần thiết: {Q:.2f} Joules")
print(f"Chi phí để đạt nhiệt độ yêu cầu: {cost:.2f} cents")
```

```python
'''
Bài 16: Viết một chương trình xác định tốc độ di chuyển của một vật khi nó chạm đất. 
Người dùng sẽ nhập chiều cao mà từ đó vật đó được thả (đơn vị tính mét (m)).
Gợi ý: Sử dụng công thức v_f =√v_i^2 + 2ad  để tính tốc độ cuối cùng, v_i tốc độ ban đầu, a là gia tốc trọng trường, và d là độ cao. 
Tốc độ ban đầu của một vật là 0 m/s,  gia tốc trọng trường là 9,8 m/s2.
'''
import math

d = float(input("Nhập chiều cao (m): "))

v_i = 0       
a = 9.8        

v_f = math.sqrt(v_i**2 + 2 * a * d)

print(f"Tốc độ khi vật chạm đất là: {v_f:.2f} m/s")
```

```python
'''
Bài 17: Khi gió thổi trong thời tiết lạnh, làm cho không khí lạnh hơn so với thực tế. 
Nguyên nhân là do sự chuyển động của không khí làm tăng tốc độ làm mát cho các vật thể (chẳng hạn, như con người).
Năm 2001, Canada, Vương quốc Anh và Hoa Kỳ đã áp dụng công thức sau đây để tính toán chỉ số gió lạnh: WCI = 13,12 + 0,6215Ta - 11,37V0.16 + 0,3965Ta V0.16
Trong công thức Ta là nhiệt độ không khí tính bằng độ C và V là tốc độ gió tính bằng km/giờ.
Viết chương trình đọc nhiệt độ không khí và tốc độ gió từ người dùng. Hiển thị chỉ số gió lạnh được làm tròn đến số nguyên gần nhất.

'''
Ta = float(input("Nhập nhiệt độ không khí (°C): "))
V = float(input("Nhập tốc độ gió (km/h): "))

WCI = 13.12 + 0.6215 * Ta - 11.37 * (V ** 0.16) + 0.3965 * Ta * (V ** 0.16)

WCI_rounded = round(WCI)

print(f"Chỉ số gió lạnh (WCI) là: {WCI_rounded}")
```
